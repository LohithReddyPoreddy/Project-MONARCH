


# Initial neuron code. Which is used to predict the value, calculate the loss, adjust the weight, update the weight, predict again and calculate the loss again




```python
w = 3
x = 2
b = 1
y = 5
lr = 0.01 # learnin rate is 0.01
for i in range(10):
  print("Epoch : ", i)
  z = (w*x + b)
  print("Predicted value is :", z)
  l = pow((z - y), 2)
  print("Loss value is : ", l)
  dw = 2*(z - y)*x
  print("Gradient with respect to weight: ", dw)
  db = 2*(z - y)
  print("Gradient with respect to bias: ", db)
  w = w - lr*(dw) #updated weight new to reduce the loss
  print("updated weight : ", w)
  b = b - lr*(db) #updated bias new to reduce the loss
  print("updated bias : ", b)
  z = (w*x + b)
  print("New Prediction : ", z)
  l = pow((z - y), 2)
  print("New Loss : ", l)
```

# Running the neuron gradient in a loop to check the loss based on the active function

```python
learning_rates = [0.01, 0.1, 0.3]
for lr in learning_rates:
    print("Learning Rate:", lr)

    w = 3
    x = 2
    b = 1
    y = 5

    for epoch in range(10):
        z = w * x + b
        l = (z - y) ** 2

        dw = 2 * (z - y) * x
        db = 2 * (z - y)

        w = w - lr * dw
        b = b - lr * db

        print("Epoch:", epoch, "Loss:", l)

    print("------------------------")
```
