## Experiment 01: Building a Neuron from Scratch

### Objective
Understand how a single artificial neuron makes predictions
and learns by adjusting its weight and bias using gradient descent.

### 1. The Neuron

A single neuron computes:

z = wx + b

Where:
- x = Input
- w = Weight
- b = Bias
- z = Predicted output

### 2. Measuring Error

Prediction error:

e = z - y

Squared error loss:

L = (z - y)²

The goal is to minimize the loss by adjusting the weight and bias.

### 3. Calculating Gradients

Using the chain rule:

dL/dw = 2(z - y)x

dL/db = 2(z - y)

These gradients tell us how the loss changes
when we change the weight or bias.

### 4. Gradient Descent

We update our parameters using:

w_new = w - learning_rate × dL/dw

b_new = b - learning_rate × dL/db

We repeat these steps until the model reaches
a sufficiently low loss or another stopping condition.




```python
w = 3
x = 2
b = 1
y = 5
lr = 0.01 # learnin rate is 0.01
for i in range(10):
  print("Epoch : ", i)
  z = (w*x + b)
  print("Predicted value is :", z)
  l = pow((z - y), 2)
  print("Loss value is : ", l)
  dw = 2*(z - y)*x
  print("Gradient with respect to weight: ", dw)
  db = 2*(z - y)
  print("Gradient with respect to bias: ", db)
  w = w - lr*(dw) #updated weight new to reduce the loss
  print("updated weight : ", w)
  b = b - lr*(db) #updated bias new to reduce the loss
  print("updated bias : ", b)
  z = (w*x + b)
  print("New Prediction : ", z)
  l = pow((z - y), 2)
  print("New Loss : ", l)
```

## Experiment 02: Understanding Learning Rates

### Objective
Observe how different learning rates affect gradient descent.

We trained the same neuron for 10 iterations using three learning rates:
0.01, 0.1, and 0.3.

### Results

| Learning Rate | Observation |
|---|---|
| 0.01 | Loss decreased gradually. |
| 0.1 | Loss reached zero after one update. |
| 0.3 | Loss increased rapidly (divergence). |

### What I Learned

The learning rate controls how much the weight and bias
change during each training iteration.

- A small learning rate can produce slow convergence.
- An appropriate learning rate can reach a minimum quickly.
- A learning rate that is too large can cause divergence.

For this particular neuron, a learning rate of 0.3 caused
the prediction to repeatedly overshoot the target.

### Next Experiment

Test learning rates of 0.15 and 0.2 to understand
oscillation and the boundary between convergence
and divergence.

```python
learning_rates = [0.01, 0.1, 0.3]
for lr in learning_rates:
    print("Learning Rate:", lr)

    w = 3
    x = 2
    b = 1
    y = 5

    for epoch in range(10):
        z = w * x + b
        l = (z - y) ** 2

        dw = 2 * (z - y) * x
        db = 2 * (z - y)

        w = w - lr * dw
        b = b - lr * db

        print("Epoch:", epoch, "Loss:", l)

    print("------------------------")
```

### Experimental Observations

**Learning rate = 0.15**
- The prediction alternates around the target.
- The error changes sign after each update.
- The magnitude of the error decreases by 50% per iteration.
- The squared error loss decreases by 75% per iteration.
- Result: Oscillating convergence.

**Learning rate = 0.2**
- The prediction alternates between 7 and 3.
- The error alternates between +2 and -2.
- The squared error loss remains constant at 4.
- Result: Oscillation without convergence.

**Learning rate = 0.3**
- The prediction crosses the target after each update.
- The magnitude of the error doubles every iteration.
- The squared error loss increases fourfold.
- Result: Divergence.

### Key Takeaway

A negative error multiplier causes the prediction error
to alternate in sign.

However, convergence depends on whether the absolute
value of that multiplier is less than 1.

For this specific experiment:

e_new = (1 - 10 × learning_rate) × e_old

Convergence occurs when:

0 < learning_rate < 0.2

```python
learning_rates = [0.01, 0.1, 0.3, 0.15, 0.2]
for lr in learning_rates:
    print("Learning Rate:", lr)

    w = 3
    x = 2
    b = 1
    y = 5

    for epoch in range(10):
        z = w * x + b
        l = (z - y) ** 2

        dw = 2 * (z - y) * x
        db = 2 * (z - y)

        w = w - lr * dw
        b = b - lr * db

        e = z - y

        print("Epoch:", epoch, "Prediction:", z, "Error:", e, "Loss:", l)

    print("------------------------")
```

## Experiment 05: Comparing Learning Rates Visually

### Objective
Compare how different learning rates affect the convergence,
oscillation, and divergence of gradient descent.

```python

learning_rates = [0.01, 0.1, 0.3, 0.15, 0.2]
all_loss_history = {}
for lr in learning_rates:
    print("Learning Rate:", lr)
    w = 3
    x = 2
    b = 1
    y = 5
    loss_history = []
    for epoch in range(10):
        z = w * x + b
        l = (z - y) ** 2

        dw = 2 * (z - y) * x
        db = 2 * (z - y)

        w = w - lr * dw
        b = b - lr * db

        e = z - y

        loss_history.append(l)

        print("Epoch:", epoch, "Prediction:", z, "Error:", e, "Loss:", l)

    print("------------------------")
    all_loss_history[lr] = loss_history
    print("Loss History for lr : ", lr, " : ", all_loss_history[lr])
print("Loss History :", all_loss_history)
```

```python
import matplotlib.pyplot as plt

for lr in learning_rates:
    plt.plot(range(10), all_loss_history[lr], label=f"lr = {lr}")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.yscale("log")
plt.title("Gradient Descent: Learning Rate Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

### Experiment 05 — Observations

We compared five learning rates using gradient descent
on a single neuron.

- lr = 0.01: Loss decreases gradually.
- lr = 0.1: Loss reaches zero after one update.
- lr = 0.15: Predictions oscillate, but loss converges to zero.
- lr = 0.2: Predictions oscillate between 7 and 3,
  keeping the loss constant at 4.
- lr = 0.3: Loss diverges rapidly.

### Key Takeaways

1. The learning rate determines the size of each gradient descent update.
2. A learning rate that is too large can cause oscillation or divergence.
3. The negative gradient gives the direction of local loss reduction,
   but does not guarantee that a large update reduces loss.
4. A logarithmic Y-axis makes it easier to compare loss curves
   with very different magnitudes.
5. Gradient descent minimizes loss, not necessarily the weight itself.

### Experiment 06 — Multiple-Input Neuron

In this experiment, I implemented a neuron with three inputs and three weights using Python.

- The prediction is calculated by multiplying each input by its corresponding weight, adding the results, and then adding the bias.
- Gradient descent updates all three weights and the bias during each epoch.
- With a learning rate of 0.01, the loss decreases from 9 and gradually approaches zero.
- The gradients are recalculated during every epoch using the current prediction error.

**Observation:** A neuron can learn by adjusting multiple weights and its bias together to reduce prediction error.

```python
# Multiple input neurons
w = [0.5,1,2]
x = [2,3,4]
b = 1
y = 10
lr = 0.01
for epoch in range(10):
  z = 0
  print("Epoch : ", epoch)
  for i in range(len(w)):
    z = z + (w[i]*x[i])
  z = z + b
  print("Predicted value is :", z)
  e = z - y
  print("Error value is : ", e)
  l = pow((z - y), 2)
  print("Loss value is : ", l)
  dw = []
  for i in range(len(w)):
    dw.append(2*(z - y)*x[i])
  print("Gradient with respect to weight: ", dw)
  db = 2*(z-y)
  print("Gradient with respect to bias: ", db)
  for i in range(len(w)):
    w[i] = w[i] - lr*(dw[i]) #updated weight new to reduce the loss
  print("updated weight : ", w)
  b = b - lr*(db) #updated bias new to reduce the loss
  print("------------------------------------------")
```

# Experiment 7 : Numpy and Vectorizations


```python
import numpy as np
w = np.array([0.5,1,2])
x = np.array([2,3,4])
b = 1
lr = 0.01
y = 10
for epoch in range(10):
  print("Epoch :", epoch)
  z = np.dot(w,x) + b
  e = z - y
  l = e**2
  dw= np.multiply(2*(z-y),x)
  db = 2*(z-y)
  w = np.subtract(w,lr*dw)
  b = b - lr*db
  print("Predicted value :", z)
  print("Error : ", e)
  print("Loss : ", l)
  print("Gradient with weights :", dw)
  print("Gradient with bias :", db)
  print("Updated weights : ", w)
  print("Updated bias: ", b)
  print("-------------------------------------")
```

```python
import numpy as np
import time

# Prepare data
n = 100_000

w = [float(i) for i in range(n)]
x = [float(i) for i in range(n)]

w_np = np.array(w, dtype=np.float64)
x_np = np.array(x, dtype=np.float64)

# 1. Pure Python implementation
start = time.perf_counter()

z_python = 0.0
for i in range(len(w)):
    z_python += w[i] * x[i]

end = time.perf_counter()
python_time = end - start

print("Python result:", z_python)
print("Python execution time:", python_time, "seconds")

# 2. NumPy implementation
start = time.perf_counter()

z_numpy = np.dot(w_np, x_np)

end = time.perf_counter()
numpy_time = end - start

print("NumPy result:", z_numpy)
print("NumPy execution time:", numpy_time, "seconds")

# 3. Compare performance
print("Speedup:", python_time / numpy_time, "x")
```
