## Experiment 01: Building a Neuron from Scratch

### Objective
Understand how a single artificial neuron makes predictions
and learns by adjusting its weight and bias using gradient descent.

### 1. The Neuron

A single neuron computes:

z = wx + b

Where:
- x = Input
- w = Weight
- b = Bias
- z = Predicted output

### 2. Measuring Error

Prediction error:

e = z - y

Squared error loss:

L = (z - y)²

The goal is to minimize the loss by adjusting the weight and bias.

### 3. Calculating Gradients

Using the chain rule:

dL/dw = 2(z - y)x

dL/db = 2(z - y)

These gradients tell us how the loss changes
when we change the weight or bias.

### 4. Gradient Descent

We update our parameters using:

w_new = w - learning_rate × dL/dw

b_new = b - learning_rate × dL/db

We repeat these steps until the model reaches
a sufficiently low loss or another stopping condition.




```python
w = 3
x = 2
b = 1
y = 5
lr = 0.01 # learnin rate is 0.01
for i in range(10):
  print("Epoch : ", i)
  z = (w*x + b)
  print("Predicted value is :", z)
  l = pow((z - y), 2)
  print("Loss value is : ", l)
  dw = 2*(z - y)*x
  print("Gradient with respect to weight: ", dw)
  db = 2*(z - y)
  print("Gradient with respect to bias: ", db)
  w = w - lr*(dw) #updated weight new to reduce the loss
  print("updated weight : ", w)
  b = b - lr*(db) #updated bias new to reduce the loss
  print("updated bias : ", b)
  z = (w*x + b)
  print("New Prediction : ", z)
  l = pow((z - y), 2)
  print("New Loss : ", l)
```

## Experiment 02: Understanding Learning Rates

### Objective
Observe how different learning rates affect gradient descent.

We trained the same neuron for 10 iterations using three learning rates:
0.01, 0.1, and 0.3.

### Results

| Learning Rate | Observation |
|---|---|
| 0.01 | Loss decreased gradually. |
| 0.1 | Loss reached zero after one update. |
| 0.3 | Loss increased rapidly (divergence). |

### What I Learned

The learning rate controls how much the weight and bias
change during each training iteration.

- A small learning rate can produce slow convergence.
- An appropriate learning rate can reach a minimum quickly.
- A learning rate that is too large can cause divergence.

For this particular neuron, a learning rate of 0.3 caused
the prediction to repeatedly overshoot the target.

### Next Experiment

Test learning rates of 0.15 and 0.2 to understand
oscillation and the boundary between convergence
and divergence.

```python
learning_rates = [0.01, 0.1, 0.3]
for lr in learning_rates:
    print("Learning Rate:", lr)

    w = 3
    x = 2
    b = 1
    y = 5

    for epoch in range(10):
        z = w * x + b
        l = (z - y) ** 2

        dw = 2 * (z - y) * x
        db = 2 * (z - y)

        w = w - lr * dw
        b = b - lr * db

        print("Epoch:", epoch, "Loss:", l)

    print("------------------------")
```
